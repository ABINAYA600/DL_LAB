```python
import tensorflow as tf
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.callbacks import EarlyStopping

# Load CIFAR-10 dataset
(X_train, y_train), (X_test, y_test) = cifar10.load_data()

# Normalize the images
X_train = X_train / 255.0
X_test = X_test / 255.0

# Convert labels to categorical format
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)


# Function to build CNN model
def build_and_compile_model(filter_size, stride, padding):

    model = Sequential()

    # First convolutional layer
    model.add(
        Conv2D(
            32,
            kernel_size=filter_size,
            strides=stride,
            padding=padding,
            activation='relu',
            input_shape=(32, 32, 3)
        )
    )

    model.add(MaxPooling2D(pool_size=(2, 2)))

    # Second convolutional layer
    model.add(
        Conv2D(
            64,
            kernel_size=filter_size,
            strides=stride,
            padding=padding,
            activation='relu'
        )
    )

    model.add(MaxPooling2D(pool_size=(2, 2)))

    # Third convolutional layer
    # Use SAME padding to avoid invalid dimensions
    # for combinations such as stride=(2,2) and padding='valid'
    model.add(
        Conv2D(
            128,
            kernel_size=filter_size,
            strides=stride,
            padding='same',
            activation='relu'
        )
    )

    model.add(Flatten())

    model.add(Dense(128, activation='relu'))

    model.add(Dense(10, activation='softmax'))

    # Compile model
    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


# Filter sizes
filter_sizes = [(3, 3), (5, 5)]

# Strides
strides = [(1, 1), (2, 2)]

# Padding
paddings = ['valid', 'same']


# Run experiments
for filter_size in filter_sizes:

    for stride in strides:

        for padding in paddings:

            print("\n======================================")
            print(
                f"Filter Size: {filter_size}, "
                f"Stride: {stride}, "
                f"Padding: {padding}"
            )
            print("======================================")

            # Build model
            model = build_and_compile_model(
                filter_size,
                stride,
                padding
            )

            # Early stopping
            early_stopping = EarlyStopping(
                monitor='val_loss',
                patience=3,
                restore_best_weights=True
            )

            # Train model
            history = model.fit(
                X_train,
                y_train,
                epochs=20,
                batch_size=64,
                validation_split=0.2,
                callbacks=[early_stopping],
                verbose=1
            )

            # Evaluate model
            test_loss, test_accuracy = model.evaluate(
                X_test,
                y_test,
                verbose=0
            )

            print(
                f"Test Accuracy: "
                f"{test_accuracy * 100:.2f}%"
            )

            # Display model summary
            model.summary()
```
