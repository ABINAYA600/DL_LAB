```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense
from sklearn.preprocessing import MinMaxScaler
import matplotlib.pyplot as plt


# Generate sine wave sequences
def generate_sine_wave(seq_length, num_samples):

    x = np.linspace(0, 2 * np.pi, seq_length)

    data = np.sin(x)

    sequences = []

    for i in range(num_samples):

        start = np.random.randint(
            0,
            len(data) - seq_length + 1
        )

        sequences.append(
            data[start:start + seq_length]
        )

    return np.array(sequences)


# Parameters
seq_length = 50
num_samples = 1000
num_epochs = 10
batch_size = 32


# Generate data
data = generate_sine_wave(
    seq_length,
    num_samples
)


# Split input and target
X = data[:, :-1]
y = data[:, -1]


# Normalize the complete data correctly
scaler = MinMaxScaler()

all_data = data.reshape(-1, 1)

scaler.fit(all_data)

X = scaler.transform(
    X.reshape(-1, 1)
).reshape(
    num_samples,
    seq_length - 1
)

y = scaler.transform(
    y.reshape(-1, 1)
).flatten()


# Reshape X for RNN
# Shape = (samples, timesteps, features)

X = X.reshape(
    (num_samples, seq_length - 1, 1)
)


# Create RNN model
model = Sequential([
    SimpleRNN(
        50,
        activation='relu',
        input_shape=(X.shape[1], 1)
    ),
    Dense(1)
])


# Compile model
model.compile(
    optimizer='adam',
    loss='mse'
)


# Display model summary
model.summary()


# Train model
history = model.fit(
    X,
    y,
    epochs=num_epochs,
    batch_size=batch_size,
    verbose=1
)


# Make predictions
predictions = model.predict(X)


# Convert predictions back to original scale
y_original = scaler.inverse_transform(
    y.reshape(-1, 1)
).flatten()

predictions_original = scaler.inverse_transform(
    predictions
).flatten()


# Plot true and predicted values
plt.figure(figsize=(12, 6))

plt.plot(
    range(len(y_original)),
    y_original,
    label='True Values'
)

plt.plot(
    range(len(predictions_original)),
    predictions_original,
    label='Predicted Values'
)

plt.title('True vs. Predicted Values')
plt.xlabel('Sample Index')
plt.ylabel('Value')

plt.legend()

plt.show()
```
