```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.preprocessing.sequence import pad_sequences


# =========================================================
# 1. Download Shakespeare text
# =========================================================

file_path = tf.keras.utils.get_file(
    "shakespeare.txt",
    origin="https://storage.googleapis.com/download.tensorflow.org/data/shakespeare.txt"
)

print("Shakespeare text downloaded successfully!")


# =========================================================
# 2. Load text
# =========================================================

def load_text(file_path):

    with open(file_path, "r", encoding="utf-8") as file:
        text = file.read().lower()

    return text


# =========================================================
# 3. Prepare data
# =========================================================

def prepare_data(text, seq_length):

    # Character-level tokenizer
    tokenizer = Tokenizer(char_level=True)

    tokenizer.fit_on_texts([text])

    # Number of unique characters
    total_chars = len(tokenizer.word_index) + 1

    sequences = []
    next_chars = []

    # Create sequences
    for i in range(0, len(text) - seq_length, 1):

        seq = text[i:i + seq_length]
        label = text[i + seq_length]

        # IMPORTANT:
        # Current Keras uses word_index even for char-level tokenizer
        sequences.append(
            [tokenizer.word_index[c] for c in seq]
        )

        next_chars.append(
            tokenizer.word_index[label]
        )

    X = np.array(sequences)
    y = np.array(next_chars)

    # Pad sequences
    X = pad_sequences(
        X,
        maxlen=seq_length
    )

    # One-hot encode input
    X = to_categorical(
        X,
        num_classes=total_chars
    )

    # One-hot encode target
    y = to_categorical(
        y,
        num_classes=total_chars
    )

    return X, y, tokenizer, total_chars


# =========================================================
# 4. Build LSTM model
# =========================================================

def build_model(seq_length, total_chars):

    model = Sequential([

        LSTM(
            128,
            input_shape=(seq_length, total_chars),
            return_sequences=True
        ),

        Dropout(0.2),

        LSTM(128),

        Dense(
            total_chars,
            activation="softmax"
        )
    ])

    model.compile(
        loss="categorical_crossentropy",
        optimizer="adam",
        metrics=["accuracy"]
    )

    return model


# =========================================================
# 5. Generate text
# =========================================================

def generate_text(
    model,
    tokenizer,
    seed_text,
    seq_length,
    total_chars,
    num_chars
):

    # Reverse character mapping
    reverse_char_index = {
        index: char
        for char, index in tokenizer.word_index.items()
    }

    generated_text = seed_text

    for _ in range(num_chars):

        # Convert characters to numbers
        sequence = [
            tokenizer.word_index.get(c, 0)
            for c in seed_text
        ]

        # Keep only last seq_length characters
        sequence = sequence[-seq_length:]

        # Pad sequence
        x_pred = pad_sequences(
            [sequence],
            maxlen=seq_length
        )

        # One-hot encoding
        x_pred = to_categorical(
            x_pred,
            num_classes=total_chars
        )

        # Predict next character
        predictions = model.predict(
            x_pred,
            verbose=0
        )[0]

        # Select character with highest probability
        next_index = np.argmax(predictions)

        next_char = reverse_char_index.get(
            next_index,
            " "
        )

        # Add character
        generated_text += next_char

        # Update seed
        seed_text = seed_text[1:] + next_char

    return generated_text


# =========================================================
# 6. Main program
# =========================================================

text = load_text(file_path)

print("Total characters:", len(text))


# Use a smaller portion so that free Colab does not
# run out of RAM during one-hot encoding.

text = text[:20000]

print("Characters used for training:", len(text))


# Sequence length
seq_length = 40


# Prepare data
X, y, tokenizer, total_chars = prepare_data(
    text,
    seq_length
)

print("Number of unique characters:", total_chars)
print("Input shape:", X.shape)
print("Output shape:", y.shape)


# =========================================================
# 7. Build model
# =========================================================

model = build_model(
    seq_length,
    total_chars
)

model.summary()


# =========================================================
# 8. Train model
# =========================================================

history = model.fit(
    X,
    y,
    epochs=20,
    batch_size=128,
    verbose=1
)


# =========================================================
# 9. Generate text
# =========================================================

seed_text = text[:seq_length]

print("\n")
print("=" * 60)
print("SEED TEXT")
print("=" * 60)

print(seed_text)


generated_text = generate_text(
    model,
    tokenizer,
    seed_text,
    seq_length,
    total_chars,
    num_chars=400
)


print("\n")
print("=" * 60)
print("GENERATED TEXT")
print("=" * 60)

print(generated_text)
```
